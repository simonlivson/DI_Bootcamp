# Week 3 Day 1 – Daily Challenge: NumPy, Pandas and Matplotlib Integration

**Scenario:** I am a data analyst working with a dataset of global weather. My task is to analyze temperature trends and visualize the results.

**Tasks**
1. Data Preparation: generate average monthly temperatures (-5 to 35 °C) for 12 months across 10 cities with NumPy, and convert them into a DataFrame with city names as index and months as columns.
2. Data Analysis: calculate the annual average temperature for each city, and find the city with the highest and lowest average.
3. Data Visualization: plot the temperature trends.
4. Report: summarize my findings.

```python
import numpy as np
# numpy for generating the data
import pandas as pd
# pandas for the table
import matplotlib.pyplot as plt
# matplotlib for the charts
```

## 1. Data Preparation

```python
# seed makes the random numbers the same every time I run the notebook
np.random.seed(42)
# np.random.uniform(-5, 35, (10, 12)) gives random temperatures between -5 and 35, 10 cities x 12 months
data = np.random.uniform(-5, 35, (10, 12))

cities = ['Tel Aviv', 'London', 'Paris', 'New York', 'Tokyo', 'Sydney', 'Moscow', 'Cairo', 'Toronto', 'Mumbai']
months = ['Jan', 'Feb', 'Mar', 'Apr', 'May', 'Jun', 'Jul', 'Aug', 'Sep', 'Oct', 'Nov', 'Dec']

# DataFrame with cities as rows (index) and months as columns
df = pd.DataFrame(data, index=cities, columns=months).round(1)
df
```

## 2. Data Analysis

```python
# mean(axis=1) takes the average across the columns, so one value per city
annual_avg = df.mean(axis=1).round(2)
annual_avg.sort_values(ascending=False)
```

```python
# idxmax() / idxmin() give the city name with the highest / lowest average
hottest = annual_avg.idxmax()
coldest = annual_avg.idxmin()
print(f'Highest average: {hottest} ({annual_avg[hottest]} °C)')
print(f'Lowest average: {coldest} ({annual_avg[coldest]} °C)')
```

## 3. Data Visualization

```python
# line plot: one line per city, months on the x axis
plt.figure(figsize=(12, 6))
for city in df.index:
    plt.plot(months, df.loc[city], marker='o', label=city)
# Another way without a loop: .T turns the cities into columns, so one plot call draws one line per city
# plt.plot(months, df.T.values, marker='o')   # one call draws all 10 lines
# plt.legend(df.index)                        # city names for the legend
plt.title('Monthly Temperatures by City')
plt.xlabel('Month')
plt.ylabel('Temperature (°C)')
plt.legend(bbox_to_anchor=(1.02, 1), loc='upper left')
plt.grid(True)
plt.tight_layout()
plt.show()
```

```python
# bar chart of the annual average for each city, sorted from hot to cold
sorted_avg = annual_avg.sort_values(ascending=False)
plt.figure(figsize=(10, 5))
plt.bar(sorted_avg.index, sorted_avg.values, color='orange')
plt.title('Annual Average Temperature by City')
plt.xlabel('City')
plt.ylabel('Average Temperature (°C)')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
# monthly average across all cities, to see if there is a trend over the year
monthly_avg = df.mean(axis=0)
plt.figure(figsize=(10, 5))
plt.plot(months, monthly_avg, marker='o', color='red')
plt.title('Average Temperature of All Cities per Month')
plt.xlabel('Month')
plt.ylabel('Temperature (°C)')
plt.grid(True)
plt.show()
print('Warmest month:', monthly_avg.idxmax(), '| Coldest month:', monthly_avg.idxmin())
```

## 4. Report: My Findings
*I set `np.random.seed(42)`, so these numbers stay the same every time I run the notebook.*

- **Highest annual average:** Cairo, about **17.6 °C**.
- **Lowest annual average:** London, about **9.6 °C**.
- The averages of all 10 cities are close to each other (about 9.6 to 17.6 °C), near the middle of the -5 to 35 range. This makes sense, because every month is a random number from the same range.
- The single hottest value is **34.5 °C** (Sydney, October) and the coldest is **-4.8 °C** (Moscow, January).
- There is **no real seasonal trend**: the all-cities average jumps up and down (warmest month August at 19.7 °C, coldest July at 8.2 °C). Real weather has summer and winter, but this random data has no seasons.
- **Conclusion:** the code works for analysis and charts, but to find real trends I would need real temperature data.
