## Exercise 1: Maximum Value Replacement in Matrix
Create a random 5x5 matrix and replace the maximum value by 0.

*The numbers are random, so my output is different from the expected output.*

```python
# 5x5 matrix of random numbers between 0 and 1
m = np.random.rand(5, 5)
print('Max value:', m.max())
# m == m.max() is True only at the biggest value, and I set it to 0
m[m == m.max()] = 0
m
```

## Exercise 2: Finding Common Values in Arrays
Find the common values between two random arrays of size 5.

Expected output: `array([0.35, 0.66, 0.76])`

```python
# two random arrays of 5 whole numbers from 0 to 9
a = np.random.randint(0, 10, 5)
b = np.random.randint(0, 10, 5)
print('a:', a)
print('b:', b)
# np.intersect1d gives the values that are in both arrays
common = np.intersect1d(a, b)
common
```

*Note: random decimal numbers almost never match, so I used random whole numbers 0-9 with `randint` so the two arrays can actually share values. The result changes each run and can be empty.*

## Exercise 3: Sorting Array in Ascending and Descending Order
Create a random 1D array of size 10 and sort it in ascending and then in descending order.

*The numbers are random, so my output is different from the expected output.*

```python
arr = np.random.rand(10)
# np.sort sorts from small to big
ascending = np.sort(arr)
# [::-1] reverses it, so it goes from big to small
descending = np.sort(arr)[::-1]
print('Ascending:', np.round(ascending, 2))
print('Descending:', np.round(descending, 2))
```

## Brief Introduction to Matrix Rank
**What is matrix rank?**
The rank of a matrix is the number of linearly independent rows or columns in the matrix. For example, in a matrix with rows `[1, 2, 3]` and `[2, 4, 6]`, the second row is just twice the first, so they are not independent and the rank is 1.

**Why and when to use it?**
I use the rank to check if a set of vectors (rows or columns) is linearly independent. This is important for understanding the solutions of systems of linear equations, and it shows if the matrix has redundant information.

## Exercise 4: Matrix Rank Determination
Generate a random 4x4 matrix and find the rank of the matrix.

Expected output: `Rank: 4`

```python
matrix = np.random.rand(4, 4)
# matrix_rank counts how many rows are linearly independent
rank = np.linalg.matrix_rank(matrix)
print(f'Rank: {rank}')
```

## Exercise 5: Creating Border Array
Create a 2D array with 1 on the border and 0 inside.

```python
# start with a 5x5 array full of 1
border = np.ones((5, 5), dtype=int)
# [1:-1, 1:-1] is everything except the first and last row and column, and I set it to 0
border[1:-1, 1:-1] = 0
border
```
