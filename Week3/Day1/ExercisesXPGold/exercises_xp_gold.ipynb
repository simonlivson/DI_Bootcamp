## Exercise 1: Minimum and Maximum of Random Array
Create a 5x5 array with random values and find the minimum and maximum values.

*The numbers are random, so my output is different from the expected output.*

````python
# 5x5 array of random numbers between 0 and 1
arr = np.random.rand(5, 5)
# min() and max() give the smallest and the biggest value
print(f'Min: {arr.min():.2f}, Max: {arr.max():.2f}')
````

## Exercise 2: Matrix Normalization
Normalize a 3x3 random matrix (subtract the mean and divide by the standard deviation of the matrix).

*The numbers are random, so my output is different from the expected output.*

````python
matrix = np.random.rand(3, 3)
# mean() is the average, std() is the standard deviation of all values
normalized = (matrix - matrix.mean()) / matrix.std()
normalized
````

## Exercise 4: Matrix Multiplication
Multiply a 5x3 random matrix by a 3x2 random matrix.

*The numbers are random, so my output is different from the expected output.*

````python
a = np.random.rand(5, 3)
b = np.random.rand(3, 2)
# @ is matrix multiplication: (5x3) @ (3x2) gives a 5x2 result
result = a @ b
print(result.shape)
result
````

## Introduction to Matrix Multiplication
In NumPy I can multiply matrices with `np.dot()` or the `@` operator. It takes the dot product of the rows of the first matrix with the columns of the second matrix.

**Rule:** the number of columns in the first matrix must equal the number of rows in the second matrix.

```python
A = np.array([[1, 2], [3, 4]])
B = np.array([[5, 6], [7, 8]])
result = A @ B
```

## Exercise 5: Matrix Multiplication
Utilize your understanding of matrix multiplication to solve this exercise. Create two matrices of compatible dimensions. Perform matrix multiplication on these matrices and display the result. Ensure the dimensions of the matrices allow for valid multiplication.

*Tip: the number of columns in the first matrix should match the number of rows in the second matrix.*

````python
# A has 2 rows and 3 columns, B has 3 rows and 2 columns
A = np.array([[1, 2, 3], [4, 5, 6]])
B = np.array([[7, 8], [9, 10], [11, 12]])
# 3 columns in A = 3 rows in B, so I can multiply them
print('A shape:', A.shape, '| B shape:', B.shape)
# @ multiplies the matrices, the result is 2x2
result = A @ B
print('Result shape:', result.shape)
result
````
