## Exercise 1: Advanced Matrix Operations
Reminder: Eigenvectors are used to understand the linear transformations of a matrix. They show the directions in which a matrix, when applied, stretches the space.

Matrix normalization is the process of adjusting the values in the matrix to a common scale, without distorting differences in the ranges of values. It’s used to bring all the variables to a uniform scale. This is particularly useful in machine learning algorithms that calculate distances between data.

Create a 5x5 matrix with random numbers and perform the following operations:
1. Calculate the eigenvalues and eigenvectors.
2. Perform matrix normalization.

*The numbers are random, so my output is different each time I run it.*

````python
# 5x5 matrix of random numbers between 0 and 1
matrix = np.random.rand(5, 5)
print(matrix.round(2))

# np.linalg.eig gives the eigenvalues and the eigenvectors (each column is one eigenvector)
eigenvalues, eigenvectors = np.linalg.eig(matrix)
print('Eigenvalues:\n', eigenvalues.round(3))
print('Eigenvectors:\n', eigenvectors.round(3))
````

*Note: a random matrix that is not symmetric can have complex eigenvalues (with a `j` part). That is normal.*

````python
# normalization: subtract the mean and divide by the standard deviation
normalized = (matrix - matrix.mean()) / matrix.std()
print(normalized.round(2))
# after normalizing, the mean is about 0 and the std is 1
print('Mean:', round(normalized.mean(), 2), '| Std:', round(normalized.std(), 2))
````

## Exercise 2: Statistical Distribution Analysis
Hint: To generate a dataset following a normal distribution, I can use numpy’s np.random.normal() function. For plotting its histogram, matplotlib’s plt.hist() can be used.

Generate a dataset following a normal distribution and plot its histogram. Then, calculate and interpret its skewness and kurtosis.

*The numbers are random, so my output is different each time I run it.*

````python
# 1000 numbers from a normal distribution with mean 0 and std 1
data = np.random.normal(0, 1, 1000)

plt.hist(data, bins=30, density=True)
plt.title('Normal Distribution')
plt.xlabel('Value')
plt.ylabel('Frequency')
plt.show()
````

````python
s = skew(data)
k = kurtosis(data)
print('Skewness:', round(s, 3))
print('Kurtosis:', round(k, 3))

# I read the numbers: close to 0 means the data looks like a normal bell
if abs(s) < 0.5:
    print('Skewness is close to 0: the histogram is almost symmetric.')
elif s > 0:
    print('Positive skew: longer tail on the right.')
else:
    print('Negative skew: longer tail on the left.')
if abs(k) < 0.5:
    print('Kurtosis is close to 0: the tails are like a normal distribution.')
elif k > 0:
    print('Positive kurtosis: more extreme values than normal.')
else:
    print('Negative kurtosis: flatter, fewer extreme values.')
````

**My interpretation:** skewness tells me if the data leans to one side. Kurtosis (scipy gives the "excess" kurtosis) tells me if the tails are heavier or lighter than a normal bell. Because I generated the data from a normal distribution, both numbers come out close to 0.

## Exercise 3: Time Series Forecasting
```python
import numpy as np

# Generating monthly sales data as a NumPy array
monthly_sales_data = np.random.randint(100, 500, size=12)  # Random sales data for 12 months
```

Using a simple NumPy array representing time series data (e.g., monthly sales data), demonstrate how to forecast future data points using linear regression.

*The numbers are random, so my output is different each time I run it.*

````python
# random sales for 12 months
monthly_sales_data = np.random.randint(100, 500, size=12)
# month numbers 1 to 12
months = np.arange(1, 13)
print(monthly_sales_data)

# polyfit with degree 1 finds the best straight line: sales = slope * month + intercept
slope, intercept = np.polyfit(months, monthly_sales_data, 1)
print('Slope:', round(slope, 2), '| Intercept:', round(intercept, 2))

# poly1d turns the slope and intercept into a function I can call
line = np.poly1d([slope, intercept])
# I forecast the next 3 months (13, 14, 15)
future_months = np.arange(13, 16)
forecast = line(future_months)
print('Forecast for months 13-15:', forecast.round(1))
````

````python
plt.plot(months, monthly_sales_data, marker='o', label='Sales')
plt.plot(months, line(months), linestyle='--', label='Trend line')
plt.plot(future_months, forecast, marker='x', color='red', label='Forecast')
plt.title('Monthly Sales and Forecast')
plt.xlabel('Month')
plt.ylabel('Sales')
plt.legend()
plt.grid(True)
plt.show()
````

*Note: the sales are random, so the line is almost flat and the forecast is not really meaningful. With real data that has a trend, the same method gives a useful forecast.*

Using a Pandas DataFrame, demonstrate grouping and aggregation techniques to summarize the data effectively.

````python
data = {
    'Product': ['Product A', 'Product B', 'Product C', 'Product A', 'Product B'],
    'Sales': [200, 150, 300, 250, 180],
    'Month': ['January', 'February', 'March', 'April', 'May']
}
df = pd.DataFrame(data)
df
````

````python
# groupby puts the rows of each product together, then sum adds their sales
print(df.groupby('Product')['Sales'].sum())
````

````python
# agg lets me do several calculations at once for each product
summary = df.groupby('Product')['Sales'].agg(['sum', 'mean', 'count', 'max', 'min'])
summary
````
