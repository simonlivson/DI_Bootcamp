## Exercise 1: Matrix Operations
In this exercise, I work with a 3x3 matrix. Here’s a brief explanation of the concepts:

- **Determinant:** The determinant is a value that can be computed from the elements of a square matrix. It provides important information about the matrix, such as whether it has an inverse, and is used in various areas like linear algebra and calculus. To understand more about it, I can watch this [video](https://www.youtube.com/watch?v=CcbyMH3Noow).
- **Inverse of a Matrix:** The inverse of a matrix is a matrix that, when multiplied with the original matrix, results in an identity matrix. Not all matrices have inverses. The inverse is crucial in solving systems of linear equations.

Create a 3x3 matrix and perform the following operations:
1. Calculate the determinant.
2. Find the inverse of the matrix.

````python
# I type a 3x3 matrix by hand
matrix = np.array([[2, 1, 3],
                   [0, 4, 1],
                   [5, 2, 6]])
print('Matrix:\n', matrix)

# np.linalg.det gives the determinant (one number)
det = np.linalg.det(matrix)
print('Determinant:', round(det, 2))
````

````python
# the determinant is not 0, so the matrix has an inverse
# np.linalg.inv gives the inverse matrix
inverse = np.linalg.inv(matrix)
print('Inverse:\n', inverse.round(3))

# check: matrix x inverse should give the identity matrix (1 on the diagonal, 0 elsewhere)
print('Matrix x Inverse:\n', np.dot(matrix, inverse).round(2))
````

## Exercise 6: Basic Hypothesis Testing
Create a sample dataset to test the effectiveness of a new training program on employee productivity:

```python
import numpy as np

# Productivity scores of employees before the training program
productivity_before = np.random.normal(loc=50, scale=10, size=30)

# Productivity scores of the same employees after the training program
productivity_after = productivity_before + np.random.normal(loc=5, scale=3, size=30)

# My task is to formulate a hypothesis regarding the training program's effectiveness 
# and test it using basic statistical functions in NumPy.
```

Given a dataset, formulate a simple hypothesis and test it using basic statistical functions in NumPy.

**My hypotheses**
- **H0 (null):** the training has no effect. The mean difference (after - before) is 0.
- **H1 (alternative):** the training increased productivity. The mean difference is bigger than 0.

I use alpha = 0.05. The same employees are measured twice, so I use a paired t-test (`ttest_rel`).

````python
# productivity scores before the training (mean 50, std 10, 30 employees)
productivity_before = np.random.normal(loc=50, scale=10, size=30)
# after = before + a random improvement around 5
productivity_after = productivity_before + np.random.normal(loc=5, scale=3, size=30)

# the difference for each employee
diff = productivity_after - productivity_before
print('Mean before:', round(np.mean(productivity_before), 2))
print('Mean after:', round(np.mean(productivity_after), 2))
print('Mean difference:', round(np.mean(diff), 2))
````

````python
# paired t-test: compares the same people before and after
# alternative='greater' because H1 says "after is bigger"
t_stat, p_value = stats.ttest_rel(productivity_after, productivity_before, alternative='greater')
print('t-statistic:', round(t_stat, 2))
print('p-value:', p_value)

alpha = 0.05
if p_value < alpha:
    print('I reject H0: the training increased productivity.')
else:
    print('I fail to reject H0: no proof that the training helped.')
````
