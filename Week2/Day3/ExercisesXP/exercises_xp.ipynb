# 🌟 Exercise 2: Descriptive Statistics
**Task:** Given a sample dataset, calculate the mean, median, variance, and standard deviation using SciPy.

**Sample Dataset:**
```python
data = [12, 15, 13, 12, 18, 20, 22, 21]
```

````python
data = [12, 15, 13, 12, 18, 20, 22, 21]
# the sample dataset from the exercise (8 numbers)
mean = stats.tmean(data)
# tmean(): the mean (average) with SciPy
median = np.median(data)
# SciPy has no separate median function, so we use NumPy (as in class)
variance = stats.tvar(data)
# tvar(): the variance with SciPy (sample variance, divides by n - 1)
std_dev = stats.tstd(data)
# tstd(): the standard deviation with SciPy (square root of tvar)
print('Mean:', mean)
# show the mean
print('Median:', median)
# show the median
print('Variance:', round(variance, 2))
# show the variance, rounded to 2 decimals
print('Standard deviation:', round(std_dev, 2))
# show the standard deviation
````

````python
print('NumPy variance (divides by n):', round(np.var(data), 2))
# NumPy's var() divides by n, so it is a bit smaller than SciPy's tvar()
print('NumPy std (divides by n):', round(np.std(data), 2))
# the same difference for the standard deviation
print('Skewness:', round(stats.skew(data), 3))
# skew(): is the data lopsided? 0 = symmetrical
print('Kurtosis:', round(stats.kurtosis(data), 3))
# kurtosis(): how heavy the tails are compared with a normal curve (0 = like a normal curve)
````

**Explanation:** For `[12, 15, 13, 12, 18, 20, 22, 21]`:
- **Mean = 16.625** (`stats.tmean`): the average value.
- **Median = 16.5**: the middle of the sorted list (between 15 and 18). It is almost the same as the mean, so the data is balanced.
- **Variance = 17.12** (`stats.tvar`) and **standard deviation = 4.14** (`stats.tstd`): on average the values are about 4 away from the mean.
- NumPy gives a smaller variance (**14.98**) and std (**3.87**) because by default it divides by n (8), while SciPy's `tvar`/`tstd` divide by n − 1 (7), the "sample" version. This is the same difference you saw in class with the daily steps.
- **Skewness = 0.065** (almost 0, so nearly symmetrical) and **kurtosis = −1.648** (negative: flat, with light tails, because the values sit in two clusters, 12–15 and 18–22, with no extreme values).

# 🌟 Exercise 3: Understanding Distributions
**Task:** Generate a normal distribution using SciPy with a mean of 50 and a standard deviation of 10. Plot this distribution.

**Code Example:**
```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import norm

# complete your code here
```

````python
from scipy.stats import norm
# norm: the normal (Gaussian, bell-shaped) distribution in SciPy
mu, sigma = 50, 10
# the mean (center) and standard deviation (spread) asked in the exercise
dist = norm(loc=mu, scale=sigma)
# create the distribution: loc = mean, scale = standard deviation
x = np.linspace(mu - 4*sigma, mu + 4*sigma, 400)
# 400 points from 10 to 90 (4 standard deviations on each side of the mean)
pdf = dist.pdf(x)
# pdf(): the height of the bell curve at each point
np.random.seed(42)
# fix the random numbers so you get the same sample as in the explanation
sample = dist.rvs(size=1000)
# rvs(): draw 1000 random values from this distribution
print('Sample mean:', round(sample.mean(), 2))
# should be close to 50
print('Sample std:', round(sample.std(ddof=1), 2))
# should be close to 10
````

````python
plt.figure(figsize=(8, 4))
# a new chart, 8 wide and 4 high
plt.hist(sample, bins=30, density=True, alpha=0.5, label='1000 random values')
# histogram of the sample; density=True puts it on the same scale as the curve
plt.plot(x, pdf, color='red', label='Normal PDF (mean 50, std 10)')
# the exact bell curve in red
plt.axvline(mu, color='black', linestyle='--', label='Mean = 50')
# a dashed line at the mean
plt.title('Normal distribution, mean = 50, std = 10')
# chart title
plt.xlabel('Value')
# label of the x axis
plt.ylabel('Probability density')
# label of the y axis
plt.legend()
# show the labels box
plt.show()
# show the chart
````

````python
print('P(value <= 60):', round(dist.cdf(60), 4))
# cdf(): the chance of a value up to 60 (1 standard deviation above the mean)
print('P(40 <= value <= 60):', round(dist.cdf(60) - dist.cdf(40), 4))
# the chance of a value within 1 standard deviation of the mean
````

**Explanation:** `norm(loc=50, scale=10)` creates the normal distribution. The red line is the exact bell curve (`pdf`), and the blue bars are 1000 random values drawn with `rvs`. The sample mean is **50.19** and its std **9.79**, very close to 50 and 10, so the sample follows the curve. The curve is symmetrical around 50, and `cdf` shows that **84.13%** of values are at most 60 and **68.27%** are between 40 and 60 (within one standard deviation of the mean).

# 🌟 Exercise 4: T-Test Application
**Task:** Perform a T-test on two sets of randomly generated data.

**Code Example:**
```python
data1 = np.random.normal(50, 10, 100)
data2 = np.random.normal(60, 10, 100)

# complete your code here
```

````python
np.random.seed(42)
# fix the random numbers so you get the same results as in the explanation (remove it to get new numbers each run)
data1 = np.random.normal(50, 10, 100)
# 100 random values around 50 with spread 10 (the code from the exercise)
data2 = np.random.normal(60, 10, 100)
# 100 random values around 60 with spread 10
print('Mean of data1:', round(data1.mean(), 2))
# the average of the first group
print('Mean of data2:', round(data2.mean(), 2))
# the average of the second group
````

````python
from scipy.stats import ttest_ind, t
# ttest_ind: the T-test for two independent groups; t: the T-distribution (for the critical value)
t_stat, p_val = ttest_ind(data1, data2)
# compare the two means: t_stat = how far apart they are, p_val = the chance of such a gap by luck alone
df_t = len(data1) + len(data2) - 2
# degrees of freedom = n1 + n2 - 2 (as in class)
critical_value = t.ppf(1 - 0.05/2, df_t)
# t.ppf(): the critical value for alpha = 0.05 (two-sided), instead of looking in a T-table
print('T-statistic:', round(t_stat, 3))
# show the T-statistic
print('P-value:', p_val)
# show the P-value
print('Degrees of freedom:', df_t, '| Critical value:', round(critical_value, 3))
# show the critical value
if abs(t_stat) > critical_value and p_val < 0.05:
    # both rules from class: |t| bigger than the critical value and p below 0.05
    print('The difference is statistically significant.')
    # the means are really different
else:
    # otherwise
    print('The difference is not statistically significant.')
    # the gap could be just chance
````

````python
plt.figure(figsize=(8, 4))
# a new chart
plt.hist(data1, bins=20, alpha=0.5, label='data1 (mean 50)')
# histogram of the first group
plt.hist(data2, bins=20, alpha=0.5, label='data2 (mean 60)')
# histogram of the second group, on the same chart
plt.title(f'T-test: t = {t_stat:.2f}, p = {p_val:.2e}')
# put the results in the title (.2e = scientific notation)
plt.xlabel('Value')
# label of the x axis
plt.ylabel('Count')
# label of the y axis
plt.legend()
# show the labels box
plt.show()
# show the chart
````

**Explanation:** The two groups have means of **48.96** and **60.22** (close to the 50 and 60 they were generated with). The T-test gives **t = −8.551** (the minus sign only means data1 is the smaller group) and **p ≈ 3.3e-15**. The critical value for 198 degrees of freedom and alpha 0.05 is **1.972** (`t.ppf`). Because |−8.551| is much bigger than 1.972 and the P-value is far below 0.05, the difference between the groups is **statistically significant**: a gap this big is practically impossible by chance alone. The chart shows it too: the two histograms overlap but are clearly shifted apart.

*Note:* we added `np.random.seed(42)` so the numbers match this explanation. Without it you get new random numbers each run, but because the real means differ by 10, the result will almost always be significant.

# 🌟 Exercise 5: Linear Regression Analysis
**Objective:** Apply linear regression to a dataset and interpret the results.

**Task:** Given a dataset of housing prices (house_prices) and their corresponding sizes (house_sizes), use linear regression to predict the price of a house given its size.

**Dataset:**
- house_sizes: [50, 70, 80, 100, 120] (in square meters)
- house_prices: [150,000, 200,000, 210,000, 250,000, 280,000] (in currency units)

**Questions:**
- What is the slope and intercept of the regression line?
- Predict the price of a house that is 90 square meters.
- Interpret the meaning of the slope in the context of housing prices.

````python
from scipy.stats import linregress
# linregress: simple linear regression (the best straight line through the points)
house_sizes = np.array([50, 70, 80, 100, 120])
# the sizes in square meters
house_prices = np.array([150000, 200000, 210000, 250000, 280000])
# the prices (written without commas, because Python reads 150,000 as two numbers)
result = linregress(house_sizes, house_prices)
# fit the line price = intercept + slope * size
print('Slope:', round(result.slope, 2))
# how much the price goes up for each extra square meter
print('Intercept:', round(result.intercept, 2))
# the price the line gives for a size of 0
print('R-squared:', round(result.rvalue**2, 4))
# R-squared: how well the line fits (1 = perfect)
print('P-value:', result.pvalue)
# the chance of such a strong line if size had no effect on price
````

````python
size_90 = 90
# the size we want to price
price_90 = result.intercept + result.slope * size_90
# use the line: intercept + slope * 90
print('Predicted price of a 90 m2 house:', round(price_90, 2))
# show the prediction
````

````python
line = result.intercept + result.slope * house_sizes
# the price the line gives for each house size
plt.figure(figsize=(7, 4))
# a new chart
plt.scatter(house_sizes, house_prices, label='Houses')
# the 5 real houses as points
plt.plot(house_sizes, line, color='red', label='Regression line')
# the regression line in red
plt.scatter([size_90], [price_90], color='green', marker='*', s=200, label='Prediction for 90 m2')
# the predicted 90 m2 house as a green star
plt.title('House price vs size')
# chart title
plt.xlabel('Size (m2)')
# label of the x axis
plt.ylabel('Price')
# label of the y axis
plt.legend()
# show the labels box
plt.show()
# show the chart
````

**Answers:**
- **Slope = 1,828.77** and **intercept = 64,383.56**. The line is: price = 64,383.56 + 1,828.77 × size.
- **A 90 m² house:** 64,383.56 + 1,828.77 × 90 = **228,972.60** currency units (the green star on the chart, between the 80 m² house at 210,000 and the 100 m² house at 250,000).
- **Meaning of the slope:** each extra square meter adds about **1,829** to the price. For example, a house 10 m² bigger is expected to cost about 18,288 more.
- The fit is very good: **R² = 0.9884**, so size explains about 98.8% of the differences in price in these 5 houses (p = 0.0005). The intercept (64,384 for 0 m²) is only where the line starts; a house of 0 m² does not exist, so it has no real meaning on its own.

# 🌟 Exercise 6: Understanding ANOVA
**Objective:** Test understanding of ANOVA and its application.

**Task:** Three different fertilizers are applied to three separate groups of plants to test their effectiveness. The growth in centimeters is recorded.

**Dataset:**
- fertilizer_1: [5, 6, 7, 6, 5]
- fertilizer_2: [7, 8, 7, 9, 8]
- fertilizer_3: [4, 5, 4, 3, 4]

**Questions:**
- Perform an ANOVA test on the given data. What are the F-value and P-value?
- Based on the P-value, do the fertilizers have significantly different effects on plant growth?
- Explain what would happen if the P-value were greater than 0.05.

````python
from scipy.stats import f_oneway
# f_oneway: one-way ANOVA, compares the means of 3 or more groups
fertilizer_1 = [5, 6, 7, 6, 5]
# growth (cm) with fertilizer 1
fertilizer_2 = [7, 8, 7, 9, 8]
# growth (cm) with fertilizer 2
fertilizer_3 = [4, 5, 4, 3, 4]
# growth (cm) with fertilizer 3
print('Means:', np.mean(fertilizer_1), np.mean(fertilizer_2), np.mean(fertilizer_3))
# the average growth of each group
f_value, p_value = f_oneway(fertilizer_1, fertilizer_2, fertilizer_3)
# run the ANOVA: F = variation between groups / variation inside groups
print('F-value:', round(f_value, 3))
# show the F-value
print('P-value:', p_value)
# show the P-value
if p_value < 0.05:
    # 0.05 is the usual limit (95% confidence)
    print('Significant: at least one fertilizer has a different effect.')
    # reject the idea that all fertilizers are the same
else:
    # otherwise
    print('Not significant: no proof that the fertilizers differ.')
    # keep the idea that all fertilizers are the same
````

````python
plt.figure(figsize=(6, 4))
# a new chart
plt.boxplot([fertilizer_1, fertilizer_2, fertilizer_3])
# one box per fertilizer: the middle line is the median
plt.xticks([1, 2, 3], ['Fertilizer 1', 'Fertilizer 2', 'Fertilizer 3'])
# names under the boxes
plt.title(f'Plant growth by fertilizer (F = {f_value:.2f}, p = {p_value:.4f})')
# results in the title
plt.ylabel('Growth (cm)')
# label of the y axis
plt.show()
# show the chart
````

**Answers:**
- **F-value = 28.526** and **P-value ≈ 0.0000275** (2.75e-05).
- **Significant?** Yes. The P-value is much smaller than 0.05, so we reject the idea that all three fertilizers work the same. The averages show the difference: fertilizer 2 = **7.8 cm**, fertilizer 1 = **5.8 cm**, fertilizer 3 = **4.0 cm**. The high F-value means the differences between the groups are much bigger than the differences inside each group. (ANOVA says that at least one group is different; it does not say which one by itself, but the boxplot shows that fertilizer 2 grows the most and fertilizer 3 the least.)
- **If the P-value were greater than 0.05:** we would **not** reject the null hypothesis. We would say there is not enough evidence that the fertilizers have different effects, because the differences we see could easily come from random variation between plants. That does not prove they are the same; it only means the data cannot show a difference.

# 🌟 Exercise 7: Probability Distributions (Optional)
Work with a binomial distribution: calculate probabilities for different numbers of successes.

**Example:** Calculating the probability of getting exactly 5 heads in 10 coin flips.

````python
from scipy.stats import binom
# binom: the binomial distribution (number of successes in n tries)
n_flips, p_heads = 10, 0.5
# 10 coin flips, chance of heads 0.5 each time
coin = binom(n_flips, p_heads)
# create the distribution
print('P(exactly 5 heads):', round(coin.pmf(5), 4))
# pmf(): the chance of exactly 5 heads
print('P(at most 3 heads):', round(coin.cdf(3), 4))
# cdf(): the chance of 0, 1, 2 or 3 heads
print('P(at least 8 heads):', round(1 - coin.cdf(7), 4))
# 1 minus the chance of 7 or fewer = the chance of 8, 9 or 10
for k in range(n_flips + 1):
    # go over every possible number of heads, 0 to 10
    print(k, 'heads:', round(coin.pmf(k), 4))
    # show the chance of exactly k heads
````

````python
k_values = np.arange(0, n_flips + 1)
# the numbers 0 to 10
plt.figure(figsize=(7, 4))
# a new chart
plt.bar(k_values, coin.pmf(k_values))
# one bar per number of heads, the height is its probability
plt.title('Binomial distribution, n = 10, p = 0.5')
# chart title
plt.xlabel('Number of heads')
# label of the x axis
plt.ylabel('Probability (PMF)')
# label of the y axis
plt.show()
# show the chart
````

**Explanation:** With `binom(10, 0.5)`:
- the chance of **exactly 5 heads is 0.2461** (about 24.6%). It is the most likely result, but it still happens less than 1 time in 4;
- the chance of **at most 3 heads is 0.1719**, and of **at least 8 heads 0.0547**;
- the chart is symmetrical around 5, because heads and tails have the same chance (p = 0.5). 0 heads or 10 heads each have only about 0.001 (0.1%).

# 🌟 Exercise 8: Correlation Coefficients (Optional)
Calculate the Pearson and Spearman correlation coefficients between two variables in a dataset.

**Example in code:**
```python
data = pd.DataFrame({'age': [23, 25, 30, 35, 40], 'income': [35000, 40000, 50000, 60000, 70000]})
```

````python
from scipy.stats import pearsonr, spearmanr
# pearsonr: linear correlation; spearmanr: rank (order) correlation
data = pd.DataFrame({'age': [23, 25, 30, 35, 40], 'income': [35000, 40000, 50000, 60000, 70000]})
# the table from the exercise
pearson_corr, pearson_p = pearsonr(data['age'], data['income'])
# Pearson: how close the points are to a straight line (-1 to 1)
spearman_corr, spearman_p = spearmanr(data['age'], data['income'])
# Spearman: do the two columns go up in the same order? (-1 to 1)
print('Pearson correlation:', round(pearson_corr, 4), '| p-value:', f'{pearson_p:.2e}')
# show Pearson and its p-value
print('Spearman correlation:', round(spearman_corr, 4))
# show Spearman
print('Covariance (age, income):', round(np.cov(data['age'], data['income'])[0, 1], 2))
# np.cov(): covariance; [0, 1] picks the age-income value from the 2x2 table
````

**Explanation:** **Pearson = 0.9997** (p ≈ 6.5e-06): age and income lie almost exactly on a straight line, a very strong positive linear relationship. **Spearman = 1.0**: the order is perfect, because every older person in the table also has a higher income. The covariance is **100,500**: positive, so the two move in the same direction, but the number itself is hard to read because it depends on the units (years × currency). That is why we use the correlation (from −1 to 1) to judge the strength. With only 5 people this is a small example, and a correlation does not prove that age causes higher income.
