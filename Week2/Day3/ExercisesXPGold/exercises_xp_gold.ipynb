# Exercise 1: Multivariate Normal Distribution
**Task:** Generate and visualize both a univariate and a multivariate normal distribution using SciPy. Explain their differences.

**Hint for Univariate Distribution:**

Use: `norm` from SciPy’s stats module.

**Code Example:**
```python
from scipy.stats import norm
univariate_data = norm.rvs(size=1000)
```
**Hint for Multivariate Distribution:**

Use: `multivariate_normal` from the same module.

**Code Example:**
```python
from scipy.stats import multivariate_normal
mean = [0, 0]
cov = [[1, 0], [0, 1]]  # diagonal covariance, independent variables
multivariate_data = multivariate_normal.rvs(mean=mean, cov=cov, size=1000)
```
**Task Explanation:** Compare the generated univariate and multivariate distributions. Discuss how the presence of multiple variables and their correlations in the multivariate case lead to differences in distribution shape and spread.

### Step 1: Univariate normal distribution (one variable)

````python
from scipy.stats import norm
# norm: the normal distribution with ONE variable
np.random.seed(0)
# fix the random numbers so you get the same results as in the explanation
univariate_data = norm.rvs(size=1000)
# 1000 random values; by default mean = 0 and std = 1 (the code from the hint)
print('Shape:', univariate_data.shape)
# 1000 values in one row: each value is a single number
print('Mean:', round(univariate_data.mean(), 3), '| Std:', round(univariate_data.std(ddof=1), 3))
# close to 0 and 1
````

````python
x = np.linspace(-4, 4, 200)
# 200 points from -4 to 4 for the curve
plt.figure(figsize=(7, 4))
# a new chart
plt.hist(univariate_data, bins=30, density=True, alpha=0.6, label='1000 random values')
# histogram of the data on the density scale
plt.plot(x, norm.pdf(x), color='red', label='Normal PDF (mean 0, std 1)')
# the exact bell curve
plt.title('Univariate normal distribution')
# chart title
plt.xlabel('Value')
# label of the x axis
plt.ylabel('Density')
# label of the y axis
plt.legend()
# show the labels box
plt.show()
# show the chart
````

### Step 2: Multivariate normal distribution (two variables)

````python
from scipy.stats import multivariate_normal
# multivariate_normal: the normal distribution with SEVERAL variables at once
mean = [0, 0]
# the center of each of the 2 variables
cov = [[1, 0], [0, 1]]  # diagonal covariance, independent variables
# covariance matrix: 1 on the diagonal = variance of each variable; 0 outside = no link between them
multivariate_data = multivariate_normal.rvs(mean=mean, cov=cov, size=1000)
# 1000 random points; each point has 2 numbers (x and y)
print('Shape:', multivariate_data.shape)
# 1000 rows and 2 columns
print('Covariance matrix of the sample:')
# the covariance measured in the data
print(np.cov(multivariate_data.T).round(3))
# np.cov wants the variables as rows, so we flip the table with .T
print('Correlation between the 2 variables:', round(stats.pearsonr(multivariate_data[:, 0], multivariate_data[:, 1])[0], 3))
# close to 0, because the variables are independent
````

To see how **correlation** changes the shape, we also make a second version where the two variables are linked (covariance 0.8).

````python
cov_corr = [[1, 0.8], [0.8, 1]]
# the same variances, but now 0.8 outside the diagonal: the variables move together
correlated_data = multivariate_normal.rvs(mean=mean, cov=cov_corr, size=1000)
# 1000 points from the correlated version
print('Correlation in the correlated sample:', round(stats.pearsonr(correlated_data[:, 0], correlated_data[:, 1])[0], 3))
# close to 0.8
````

````python
xx, yy = np.mgrid[-4:4:0.05, -4:4:0.05]
# a grid of points to draw the density contours (the "height lines" of the 2D bell)
grid = np.dstack((xx, yy))
# stack x and y so each grid point is a pair (x, y)
fig, axes = plt.subplots(1, 2, figsize=(12, 5))
# two charts side by side
for ax, pts, c, title in [(axes[0], multivariate_data, cov, 'Independent (cov = 0)'), (axes[1], correlated_data, cov_corr, 'Correlated (cov = 0.8)')]:
    # draw the same chart for both versions
    ax.scatter(pts[:, 0], pts[:, 1], s=5, alpha=0.4)
    # each random point as a small dot
    ax.contour(xx, yy, multivariate_normal(mean, c).pdf(grid), levels=6, colors='red')
    # red rings of equal density (pdf of the 2D distribution)
    ax.set_title('Multivariate normal: ' + title)
    # chart title
    ax.set_xlabel('Variable 1')
    # label of the x axis
    ax.set_ylabel('Variable 2')
    # label of the y axis
    ax.set_aspect('equal')
    # same scale on both axes, so a circle looks like a circle
plt.tight_layout()
# leave space between the charts
plt.show()
# show the charts
````

**Comparison and differences:**
- **Univariate (one variable):** `norm.rvs(size=1000)` gives 1000 single numbers (shape **(1000,)**). The sample mean is **−0.045** and the std **0.988**, close to the true 0 and 1. Its picture is the familiar **bell curve** on one axis, and its spread is described by **one number**, the standard deviation.
- **Multivariate (two variables):** `multivariate_normal.rvs` gives 1000 **points**, each with 2 numbers (shape **(1000, 2)**). Its picture is a **cloud of points** on a flat plane, and the density is a "2D bell" (a hill) whose height lines are the red rings. Its spread is described by a **covariance matrix**, not one number: the diagonal holds the variance of each variable (sample: **0.951** and **0.900**, close to 1) and the off-diagonal holds how the two move together (sample: **0.016**, close to 0).
- **Effect of correlation:** with `cov = [[1, 0], [0, 1]]` the variables are independent (measured correlation **0.018**), so the cloud is **round** and the rings are circles: knowing variable 1 tells you nothing about variable 2. When we put 0.8 outside the diagonal (measured correlation **0.799**), the cloud becomes a **tilted ellipse** along the diagonal: when one variable is high, the other usually is too. Each variable on its own still has the same bell shape and std 1; only the **joint** shape changes. So in the multivariate case the shape and spread depend not only on each variable's variance but also on the correlations between them.

# Exercise 2: Advanced Probability Distributions
**Task:** Choose an advanced probability distribution (e.g., Poisson, exponential) and demonstrate (in text, not code) its application in a real-world scenario.

**Real-World Scenarios to Choose From:**
- **Poisson Distribution:** Use this to model the number of customer arrivals at a store in a given time interval.
- **Exponential Distribution:** Exponential Distribution: Apply this to model the time until the next earthquake in an earthquake-prone region.

**Explanation:** Describe how the chosen distribution applies to the scenario, including the assumptions and implications of the distribution in the context.

**My choice: the Poisson distribution, for customer arrivals at a store.**

**The scenario.** A small shop gets on average **λ = 12 customers per hour** at midday. The Poisson distribution gives the probability of seeing exactly k customers in one hour: P(k) = λᵏ · e^(−λ) / k!. With λ = 12, the most likely counts are around 11–12 customers, but 6 or 18 in an hour also happen sometimes. The mean and the variance are both λ (12), so the standard deviation is √12 ≈ 3.5 customers.

**How it applies.** We count **events** (arrivals) in a **fixed interval** (one hour). Each customer decides to come on their own, and many people could come but each one has only a small chance of coming in any given minute. That is exactly the situation the Poisson distribution describes.

**Assumptions.**
- **Independence:** one customer's arrival does not change the chance of another arriving (this breaks when people come in groups, such as a family or a tour bus).
- **Constant rate:** λ stays the same during the interval. In reality, a lunch rush is busier than 3 PM, so we use a separate λ for each time slot.
- **One at a time:** two arrivals do not happen at exactly the same instant.
- **Mean = variance:** if the real counts vary much more than the mean (for example, quiet days and very busy sale days mixed together), the plain Poisson model underestimates the spread.

**Implications.**
- **Staffing:** the manager can compute the chance that more than, say, 18 customers come in an hour (about 4% when λ = 12) and plan enough cashiers for busy hours instead of only the average.
- **Queues and stock:** the same counts help to plan waiting times at the checkout and how much fresh food to prepare.
- **Spotting changes:** if a promotion brings 25 customers in an hour, the model says this is very unlikely under λ = 12, so the promotion probably raised the rate.
- **Link to the exponential distribution:** if arrivals follow a Poisson process with 12 per hour, the **time between** two customers follows an exponential distribution with an average of 5 minutes. The two distributions are two sides of the same process.

# Exercise 3: Statistical Hypothesis Testing
**Task:** Perform and interpret an ANOVA test using SciPy to compare more than two groups.

**Groups to Use:** You can use the data provided from the following code:
```python
import numpy as np
import pandas as pd

# Seed for reproducibility
np.random.seed(0)
# Sales data across different regions
region1 = np.random.normal(20000, 3000, 30)  # Region 1 sales
region2 = np.random.normal(22000, 3500, 30)  # Region 2 sales
region3 = np.random.normal(25000, 5000, 30)  # Region 3 sales
sales_data = pd.DataFrame({
    'Region 1': region1,
    'Region 2': region2,
    'Region 3': region3
})
```
**Additional Resource:** Here’s a link to a [guide on ANOVA test](https://docs.scipy.org/doc/scipy/reference/generated/scipy.stats.f_oneway.html). This will help you understand the test, as it has not been covered in your lessons yet.

````python
np.random.seed(0)
# Seed for reproducibility (the same random numbers every run)
region1 = np.random.normal(20000, 3000, 30)  # Region 1 sales
# 30 sales values around 20,000 with spread 3,000
region2 = np.random.normal(22000, 3500, 30)  # Region 2 sales
# 30 sales values around 22,000 with spread 3,500
region3 = np.random.normal(25000, 5000, 30)  # Region 3 sales
# 30 sales values around 25,000 with spread 5,000
sales_data = pd.DataFrame({
    'Region 1': region1,
    'Region 2': region2,
    'Region 3': region3
})
# put the three regions in one table, one column per region
print(sales_data.describe().round(0))
# count, mean, std, min, quartiles and max for each region
````

````python
from scipy.stats import f_oneway
# f_oneway: one-way ANOVA (compares the means of 3 or more groups)
f_value, p_value = f_oneway(sales_data['Region 1'], sales_data['Region 2'], sales_data['Region 3'])
# H0 (null hypothesis): all 3 regions have the same mean sales
print('F-value:', round(f_value, 3))
# variation between the regions / variation inside the regions
print('P-value:', p_value)
# the chance of such an F-value if all regions were really the same
if p_value < 0.05:
    # 0.05 = the usual 5% limit
    print('Reject H0: at least one region has different average sales.')
    # the difference is significant
else:
    # otherwise
    print('Do not reject H0: no proof that the regions differ.')
    # the difference may be chance
````

````python
from scipy.stats import ttest_ind
# after a significant ANOVA we check the pairs with T-tests, to see which regions differ
pairs = [('Region 1', 'Region 2'), ('Region 1', 'Region 3'), ('Region 2', 'Region 3')]
# the 3 possible pairs
for a, b in pairs:
    # go over each pair
    t_stat, p = ttest_ind(sales_data[a], sales_data[b])
    # T-test for the two regions
    print(f'{a} vs {b}: t = {t_stat:.3f}, p = {p:.4f}')
    # show the result
````

````python
plt.figure(figsize=(7, 4))
# a new chart
plt.boxplot([sales_data['Region 1'], sales_data['Region 2'], sales_data['Region 3']])
# one box per region (middle line = median)
plt.xticks([1, 2, 3], sales_data.columns)
# region names under the boxes
plt.title(f'Sales by region (ANOVA F = {f_value:.2f}, p = {p_value:.4f})')
# results in the title
plt.ylabel('Sales')
# label of the y axis
plt.show()
# show the chart
````

**Interpretation:**
- The table shows the sample means: **Region 1 ≈ 21,329**, **Region 2 ≈ 20,987**, **Region 3 ≈ 24,331**. (Region 1 came out a little above Region 2 in this sample, even though it was generated around 20,000 and Region 2 around 22,000. That is random variation with only 30 values each.)
- **ANOVA:** **F = 6.862**, **p = 0.0017**. The P-value is below 0.05, so we **reject H0** (that all regions have the same mean sales): at least one region is different.
- **Which one?** ANOVA alone does not say. The pairwise T-tests show that **Region 3 differs from Region 1 (p = 0.0067) and from Region 2 (p = 0.0025)**, but **Region 1 and Region 2 do not differ (p = 0.685)**. So Region 3 is the one with clearly higher sales, which the boxplot also shows (its box sits higher and is wider, because its spread is also bigger: std ≈ 4,825).
- *Note:* testing several pairs raises the chance of a false alarm; a stricter limit such as 0.05 / 3 ≈ 0.017 (Bonferroni) still leaves both Region 3 results significant.

# Exercise 4: Linear Regression Analysis
**Task:** Implement a simple linear regression analysis using SciPy on the provided dataset and interpret the results.

**Dataset Generation Code:**
```python
import numpy as np
import pandas as pd
from scipy.stats import linregress

# Seed for reproducibility
np.random.seed(0)

# Sample data
X = np.random.rand(100) * 50  # Independent variable (e.g., hours studied)
Y = 2.5 * X + np.random.randn(100) * 10  # Dependent variable (e.g., test scores)

# Creating a DataFrame
linear_regression_data = pd.DataFrame({
  'Hours Studied': X,
  'Test Score': Y
})
```
**Steps to Follow:**
- **Data Preparation:**
  - Dependent Variable (Y): ‘Test Score’
  - Independent Variable (X): ‘Hours Studied’
- **Performing Regression:** Use the `linregress` function from SciPy’s stats module.
- **Interpretation:** Focus on the slope, intercept, and R-squared values to understand the relationship between study hours and test scores.

### Data generation (the code from the exercise)

````python
from scipy.stats import linregress
# linregress: simple linear regression
np.random.seed(0)
# Seed for reproducibility
X = np.random.rand(100) * 50  # Independent variable (e.g., hours studied)
# 100 random study times between 0 and 50 hours
Y = 2.5 * X + np.random.randn(100) * 10  # Dependent variable (e.g., test scores)
# the score is 2.5 points per hour plus random noise (spread 10)
linear_regression_data = pd.DataFrame({
  'Hours Studied': X,
  'Test Score': Y
})
# put both in one table
print(linear_regression_data.head())
# the first 5 rows
````

### Step 1: Data preparation

````python
x = linear_regression_data['Hours Studied']
# independent variable (X): what we use to predict
y = linear_regression_data['Test Score']
# dependent variable (Y): what we want to predict
print('Rows:', len(x), '| Missing values:', linear_regression_data.isnull().sum().sum())
# 100 rows, and we check there are no empty cells
````

### Step 2: Performing the regression

````python
result = linregress(x, y)
# fit the line: Test Score = intercept + slope * Hours Studied
print('Slope:', round(result.slope, 4))
# extra points per extra hour of study
print('Intercept:', round(result.intercept, 4))
# the score the line gives for 0 hours
print('R-squared:', round(result.rvalue**2, 4))
# share of the differences in scores explained by study hours
print('Correlation r:', round(result.rvalue, 4))
# Pearson correlation between hours and score
print('P-value:', result.pvalue)
# the chance of such a slope if hours had no effect
print('Standard error of the slope:', round(result.stderr, 4))
# how precise the slope estimate is
````

````python
order = np.argsort(x.values)
# the order of the rows from fewest to most hours (to draw the line left to right)
plt.figure(figsize=(8, 5))
# a new chart
plt.scatter(x, y, alpha=0.6, label='Students')
# each student as a point
plt.plot(x.values[order], result.intercept + result.slope * x.values[order], color='red', label=f'y = {result.intercept:.2f} + {result.slope:.2f}x')
# the regression line in red, with its formula in the legend
plt.title(f'Test score vs hours studied (R² = {result.rvalue**2:.3f})')
# chart title
plt.xlabel('Hours Studied')
# label of the x axis
plt.ylabel('Test Score')
# label of the y axis
plt.legend()
# show the labels box
plt.show()
# show the chart
````

### Step 3: Interpretation

**Interpretation:**
- **Slope = 2.4874:** each extra hour of study adds about **2.49 points** to the test score. This is very close to the 2.5 used to generate the data, so the regression found the real relationship.
- **Intercept = 2.2215:** the predicted score for a student who studied 0 hours. The true value used to generate the data is 0; 2.22 is small compared with the noise (spread 10), so it is close.
- **R-squared = 0.9283:** about **93%** of the differences in test scores are explained by the hours studied; the other 7% is the random noise. The correlation is **r = 0.9635**, a very strong positive relationship.
- **P-value ≈ 6.8e-58** (and slope standard error 0.07): the slope is far from 0, so the link between hours and scores is statistically significant.
- **In plain words:** more study hours go with clearly higher scores, and a straight line describes this very well. For example, 20 hours predicts about 2.22 + 2.49 × 20 ≈ 52 points.
