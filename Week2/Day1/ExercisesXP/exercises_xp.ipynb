## Exercise 1: Introduction to Data Analysis

**What is data analysis?**
Data analysis is the process of collecting, cleaning, organizing and studying data to find useful information, patterns and answers to questions. It turns raw numbers and facts into knowledge that people can use to make decisions.

**Why is data analysis important today?**
Today almost everything we do creates data: online shopping, phones, apps, sensors and social media. Companies and governments have huge amounts of it. Data analysis helps them make decisions based on facts instead of guesses, find problems early, save money and predict what may happen next. Trends like AI and machine learning, real-time dashboards and big data tools make analysis faster and more important every year.

**Three areas where data analysis is used today:**
1. **Business and retail:** stores study sales and customer data to decide which products to stock, set prices and make personal recommendations (like Amazon's "customers also bought").
2. **Healthcare:** hospitals study patient data to find diseases early, predict which patients are at risk and choose the best treatments.
3. **Finance:** banks study transactions to detect fraud and decide who gets a loan or a credit card (like the Credit Card Approvals dataset in Exercise 2).

## Exercise 3: Identifying Data Types
- **Quantitative** = numbers we can measure or count and do math with (average, sum).
- **Qualitative** = categories or labels that describe a group, even when they are written as numbers (like 1/0 for yes/no).

> Note: Colab shows text columns as `object`, so the code uses `include='object'` to pick them.

```python
print(sleep_df.dtypes)
# the type of every column in the sleep dataset
print(sleep_df.select_dtypes(include='number').columns)
# columns stored as numbers
print(sleep_df.select_dtypes(include='object').columns)
# columns stored as text
```

**Sleep dataset:**
| Column | Type | Why |
|---|---|---|
| index | Qualitative (ID) | just the row number, a label and not a measurement |
| Year | Quantitative | a number on a time scale (2003–2017); we can find the difference between years |
| Period | Qualitative | a label ("Annual") |
| Avg hrs per day sleeping | Quantitative | a measured amount of hours |
| Standard Error | Quantitative | a measured number that shows how accurate the average is |
| Type of Days | Qualitative | categories (All days, Nonholiday weekdays, Weekend days and holidays) |
| Age Group | Qualitative | categories (age ranges like "25 to 34 years") |
| Activity | Qualitative | a label ("Sleeping") |
| Sex | Qualitative | categories (Both, Men, Women) |

## Exercise 5: Basic Observation Skills (Sleep dataset)
We use the `sleep_df` we loaded in Exercise 2 (after a restart, run the Exercise 2 cells again first, so `sleep_df` exists).

```python
print(sleep_df.columns)
# show all the column names
print(sleep_df.describe())
# basic statistics of the number columns
```

The `describe()` above only shows number columns. To see what is inside the text columns, we print every different value in each one with `.unique()` (it lists each value once).

```python
print(sleep_df['Period'].unique())
# every different value in Period
print(sleep_df['Activity'].unique())
# every different value in Activity
print(sleep_df['Sex'].unique())
# every different value in Sex
print(sleep_df['Age Group'].unique())
# every different value in Age Group
print(sleep_df['Type of Days'].unique())
# every different value in Type of Days
```

**How we decide, using the outputs above:**

**Trend analysis:** `Year` + `Avg hrs per day sleeping`.
- In `describe()`, `Year` has min 2003 and max 2017, so we have 15 years of data.
- `Avg hrs per day sleeping` goes from min 7.72 to max 10.65, so the value changes.
- Together they let us check if Americans sleep more or less over time.

**Group comparison:** the `.unique()` output shows each of these columns has several groups:
- `Sex` shows Both, Men, Women: we can compare how much men and women sleep.
- `Age Group` shows 7 age groups (15 to 24 years ... 65 years and over): we can compare young and older people.
- `Type of Days` shows All days, Nonholiday weekdays, Weekend days and holidays: we can compare weekdays with weekends.

**Not useful:** the `.unique()` output shows `Period` has only 'Annual' and `Activity` has only 'Sleeping'. A column with one value is the same in every row, so it can't show any difference.

**Helper column:** `Standard Error` (min 0.018, max 0.228 in `describe()`) is small next to the sleep hours, so the averages are fairly reliable.
