**Boxplot for numerical values**: a boxplot shows the five-number summary and helps to find outliers with the IQR method.

```python
plt.figure(figsize=(10,6))
# new chart
sns.boxplot(x=df.charges)
# boxplot of charges
plt.title('Distribution Charges', size=18)
# chart title
plt.show()
# draw the chart
```

```python
Q1 = df['charges'].quantile(0.25)
# first quartile (25%)
Q3 = df['charges'].quantile(0.75)
# third quartile (75%)
IQR = Q3 - Q1
# inter quartile range
print(IQR)
# show the IQR
```

```python
outliers = df[(df['charges'] < Q1 - 1.5 * IQR) | (df['charges'] > Q3 + 1.5 * IQR)]
# rows below Q1 - 1.5*IQR or above Q3 + 1.5*IQR are outliers
print(len(outliers))
# how many outliers
outliers.head()
# show the first few outliers
```

The IQR is **11,911.37** and there are **139 outlier rows**, all with very high charges. Like the post, we keep them: before removing outliers you should check with someone who knows the domain whether they are real.

## Step 3: Data Cleaning
- **Cabin**: most values are missing, so we remove the column.
- **Age**: we fill the missing ages with the median age.
- **Embarked**: only 2 are missing, so we fill them with the most common port.
- **Sex** and **Embarked** are text, so we add number versions of them.

```python
titanic = titanic.drop(columns=['Cabin'])
# remove the Cabin column
print(titanic['Age'].median())
# the median age
titanic['Age'] = titanic['Age'].fillna(titanic['Age'].median())
# fill missing ages with the median age
print(titanic['Embarked'].mode()[0])
# the most common port
titanic['Embarked'] = titanic['Embarked'].fillna(titanic['Embarked'].mode()[0])
# fill missing ports with the most common port
titanic.isnull().sum()
# check: no missing values left
```

```python
titanic['Sex_num'] = titanic['Sex'].map({'male': 0, 'female': 1})
# new column: male = 0, female = 1
titanic['Embarked_num'] = titanic['Embarked'].map({'S': 0, 'C': 1, 'Q': 2})
# new column: Southampton = 0, Cherbourg = 1, Queenstown = 2
titanic[['Sex', 'Sex_num', 'Embarked', 'Embarked_num']].head()
# check the new columns next to the old ones
```

The median age is **28** and the most common port is **S** (Southampton). We keep the text columns too, because they make the charts easier to read.

- **Age**: the two boxplots look similar, but children under 16 survived more (59%) than passengers 16 and older (36%).

```python
print(titanic.groupby(['Pclass', 'Sex'])['Survived'].mean())
# survival rate for each class and sex together
sns.barplot(x='Pclass', y='Survived', hue='Sex', data=titanic)
# survival rate per class, split by sex
plt.title('Survival rate by class and sex')
# chart title
plt.show()
# draw the chart
```

```python
sns.scatterplot(x='Age', y='Fare', hue='Survived', data=titanic)
# age vs fare, colored by survival
plt.title('Age, fare and survival')
# chart title
plt.show()
# draw the chart
print(titanic.groupby('Survived')['Fare'].median())
# median fare of those who died (0) and survived (1)
```

```python
print(titanic.groupby('Embarked')['Survived'].mean())
# survival rate for each port
```

- Class and sex together: women in 1st class (97%) and 2nd class (92%) almost all survived. Women in 3rd class had only 50%. Men in 2nd (16%) and 3rd class (14%) had the lowest survival.
- In the scatter plot, the low-fare area (bottom) has many non-survivors, while passengers with high fares survived more often. The median fare of survivors was 26, compared to 10.5 for those who died.

## Step 5: Insight Generation
**Key findings**
- **Higher class = more likely to survive**: 63% (1st) vs 47% (2nd) vs 24% (3rd).

### Sentiment analysis with TextBlob
TextBlob gives each text a **polarity** from -1 (very negative) to +1 (very positive).

```python
!pip install textblob
# install TextBlob (if it is not installed already)
```

```python
from textblob import TextBlob
# import TextBlob
def get_polarity(text):
# a function that gets the polarity of one text
    return TextBlob(text).sentiment.polarity
    # number from -1 (negative) to +1 (positive)
tickets['polarity'] = tickets['Ticket Description'].apply(get_polarity)
# polarity for every ticket description
```

```python
def get_label(p):
# a function that turns the polarity into a word
    if p > 0:
    # above 0
        return 'positive'
        # positive text
    elif p < 0:
    # below 0
        return 'negative'
        # negative text
    else:
    # exactly 0
        return 'neutral'
        # neutral text
tickets['sentiment'] = tickets['polarity'].apply(get_label)
# label every ticket
print(tickets['sentiment'].value_counts())
# number of tickets per label
print(tickets['polarity'].mean())
# average polarity
```

```python
sns.countplot(x='sentiment', data=tickets)
# bar chart of the sentiment labels
plt.title('Ticket description sentiment (TextBlob)')
# chart title
plt.show()
# draw the chart
```

```python
tickets.groupby('sentiment')['Customer Satisfaction Rating'].mean()
# average satisfaction rating for each sentiment label (only closed tickets have a rating)
```

**Ticket findings:**
- TextBlob labels 4403 tickets positive, 2446 negative and 1620 neutral. The average polarity is only 0.06, so most texts are close to neutral.
- This is surprising for support tickets, which are about problems. Polite phrases like "Please assist" or "I'd love" push the score up, and TextBlob does not understand the problem behind the words.
- The satisfaction rating is almost the same for every label (2.95 negative, 3.07 neutral, 2.98 positive), so the text sentiment does not predict satisfaction in this dataset.

## Challenges with the unstructured data (compared to the structured data)
- **No ready-made numbers:** with sales we could sum and plot right away. With tickets we first had to turn the text into numbers (polarity) before we could analyze it.
- **Messy and template text:** every description has the `{product_purchased}` placeholder, and many sentences are mixed or cut, so the text is not reliable.
- **Sentiment tools are not accurate:** TextBlob counts words. It scored many problem reports as positive because of polite words.
- **Hard to check:** in the sales data we could check a number (like the unit price). For text there is no simple way to know if the sentiment label is right without reading tickets one by one.
- **Slower:** running TextBlob on 8469 texts takes much longer than summing 4600 rows of numbers.
- The structured data had problems too (26 impossible dates), but they were easy to find with `pd.to_datetime(errors='coerce')`.

## Tools and techniques that worked
**Structured data (sales):** pandas `read_csv`, `info`, `describe`, `sum`, `groupby`, `pd.to_datetime` for dates, and simple bar and line charts.

**Unstructured data (tickets):** reading samples of the text, `str.contains` to find patterns, **TextBlob** to turn the text into a polarity number and a label, and then the structured tools (`value_counts`, `groupby`, `countplot`) on the new columns. The ticket's structured columns (type, priority, channel, rating) were also very useful.

# Exercise 5: Converting Unstructured Data to Structured Data
**How to load (Colab):** click [Twitter Tweets Sentiment Dataset.zip](https://github.com/devtlv/Datasets-DA-Bootcamp-2-/raw/refs/heads/main/Week%204%20-%20Data%20Understanding/W4D3%20-%20Importing%20Data,%20Exporting%20D/Twitter%20Tweets%20Sentiment%20Dataset.zip) to download it, then run the upload cell and choose the zip. It contains the folder `Twitter Tweets Sentiment Dataset` with `Tweets.csv`.

```python
tweets = pd.read_csv('Twitter Tweets Sentiment Dataset/Tweets.csv')
# load the tweets into a DataFrame
tweets.head()
# first 5 rows
```

```python
tweets.info()
# columns, data types and non-null counts
```

27481 tweets with 4 columns: `textID`, `text` (the tweet), `selected_text` (the part of the tweet that shows the sentiment) and `sentiment`. One tweet has no text, so we remove it.

```python
tweets = tweets.dropna(subset=['text'])
# remove the row with no text
print(len(tweets))
# number of tweets left
```

## Step 1: Extract key information with text processing
We use `str.findall` with a simple pattern:
- `#\w+` finds hashtags (a # followed by letters, numbers or _)
- `@\w+` finds mentions (an @ followed by letters, numbers or _)
- `http\S+` finds links (http followed by anything that is not a space)

```python
tweets['hashtags'] = tweets['text'].str.findall(r'#\w+')
# list of hashtags in each tweet
tweets['mentions'] = tweets['text'].str.findall(r'@\w+')
# list of mentions in each tweet
tweets['links'] = tweets['text'].str.findall(r'http\S+')
# list of links in each tweet
```

## Step 2: Organize the information in a structured table

```python
tweets['num_hashtags'] = tweets['hashtags'].str.len()
# how many hashtags each tweet has
tweets['num_mentions'] = tweets['mentions'].str.len()
# how many mentions each tweet has
tweets['num_links'] = tweets['links'].str.len()
# how many links each tweet has
tweets['num_words'] = tweets['text'].str.split().str.len()
# how many words each tweet has
structured = tweets[['textID', 'sentiment', 'hashtags', 'mentions', 'links', 'num_hashtags', 'num_mentions', 'num_links', 'num_words']]
# the new structured table: one column for each type of information
structured.head()
# first 5 rows
```

```python
structured[structured['num_hashtags'] > 0].head()
# some tweets that have hashtags
```

## Step 3: Basic analysis of the structured data

```python
print(structured['sentiment'].value_counts())
# number of tweets per sentiment
print((structured['num_hashtags'] > 0).sum())
# number of tweets with at least one hashtag
print((structured['num_mentions'] > 0).sum())
# number of tweets with at least one mention
print((structured['num_links'] > 0).sum())
# number of tweets with at least one link
```

```python
structured.groupby('sentiment')[['num_hashtags', 'num_mentions', 'num_links', 'num_words']].mean()
# average of each count for each sentiment
```

```python
all_hashtags = structured['hashtags'].explode().str.lower()
# put every hashtag on its own row, in lower case
all_hashtags.value_counts().head(10)
# the 10 most used hashtags
```

```python
all_mentions = structured['mentions'].explode().str.lower()
# put every mention on its own row, in lower case
all_mentions.value_counts().head(10)
# the 10 most used mentions
```

```python
sns.countplot(x='sentiment', data=structured)
# bar chart of tweets per sentiment
plt.title('Tweets per sentiment')
# chart title
plt.show()
# draw the chart
```

```python
structured.groupby('sentiment')['num_words'].mean().plot(kind='bar')
# bar chart of the average number of words per sentiment
plt.title('Average words per tweet by sentiment')
# chart title
plt.show()
# draw the chart
```

**Findings from the structured table:**
- 11117 neutral, 8582 positive and 7781 negative tweets.
- Hashtags are rare: only **523 tweets** (about 2%) have a hashtag. The most common are `#fb`, `#followfriday`, `#bgt` and `#ff`.
- Mentions are almost missing: only **74 tweets** have one, and they all start with `@_`. It looks like this dataset removed the user names from the tweets, so a mention is only kept when the name started with `_`.
- **1223 tweets** have a link.
- The number of hashtags and mentions is about the same for every sentiment, so they don't tell us the sentiment. Negative tweets are slightly longer (13.5 words) than positive (13.1) and neutral (12.3).

## Comparison with the original unstructured format
- **Original text:** to know how many tweets use hashtags or which hashtags are popular, we would have to read 27,480 tweets one by one. We can't sort, count or chart the text directly.
- **Structured table:** each type of information has its own column, so with one line of pandas we can count, group by sentiment, find the top hashtags and draw charts.
- **What we lose:** the structured table keeps only what we extracted. The meaning, jokes, emotions and context of the tweet stay in the text. That's why the `sentiment` column (made from the text) is still the most useful column.
- **What we discovered only after structuring:** hashtags and mentions are very rare in this dataset and are not linked to sentiment, and most mentions were removed. This was not visible from reading a few tweets.
