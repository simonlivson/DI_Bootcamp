## Step 3: Import the libraries

````python
from PIL import Image, ImageOps
# Image opens pictures, ImageOps flips and mirrors them
import matplotlib.pyplot as plt
# matplotlib shows the pictures
from scipy.ndimage import rotate
# rotate turns a picture by some degrees
import numpy as np
# numpy turns a picture into an array of numbers
````

## Step 4: Load an image and display it

````python
image_path = 'Flower Color Images Dataset/flowers/flowers/19_010.png'
# where the image is (the course's path inside the GitHub zip folder)
original_image = Image.open(image_path)
# load the image using PIL
plt.imshow(original_image)
# display the original image using matplotlib
````

## Step 5: Rotate the image by 30 degrees

````python
def rotate_image_30_degrees(image):
# a function that gets an image
    return rotate(image, 30, reshape=False, mode='nearest')
    # turn it 30 degrees, keep the same size, and fill the corners with the nearest colors
````

````python
image_array = np.array(original_image)
# turn the PIL image into a numpy array, because rotate works on arrays
rotated_image = rotate_image_30_degrees(image_array)
# rotate the image 30 degrees with our function
plt.imshow(rotated_image)
# display the rotated image
````

## Step 6: Flip the image horizontally and then vertically

````python
mirrored_image = ImageOps.mirror(original_image)
# flip the image horizontally (left becomes right)
plt.imshow(mirrored_image)
# display the horizontally flipped image
````

````python
flipped_image = ImageOps.flip(mirrored_image)
# then flip the mirrored image vertically (top becomes bottom)
plt.imshow(flipped_image)
# display the image flipped horizontally and then vertically
````

## Step 7: Zoom in on the image (scale by 1.2x)

````python
width, height = original_image.size
# get the width and height of the image
zoomed_image = original_image.resize((int(width * 1.2), int(height * 1.2)))
# make the image 1.2 times bigger
print(original_image.size, zoomed_image.size)
# compare the old size and the new size
plt.imshow(zoomed_image)
# display the zoomed image
````
