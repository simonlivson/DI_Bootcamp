```python
sales_df = pd.read_csv('Retail Data Analytics dataset/sales data-set.csv')
# weekly sales for each store and department
stores_df = pd.read_csv('Retail Data Analytics dataset/stores data-set.csv')
# the type and size of each of the 45 stores
print(sales_df.head())
# show the first 5 sales rows
print(stores_df.head())
# show the first 5 store rows
print(sales_df.describe())
# basic statistics of the number columns
```

#### Sales trends
The dates are text written day/month/year (like `05/02/2010`), so we turn them into real dates first.
- `pd.to_datetime(column, format='%d/%m/%Y')` reads text as a date; `format` tells pandas the order: day (`%d`), month (`%m`), 4-digit year (`%Y`).
- `.dt.year` takes only the year from each date.

```python
sales_df['Date'] = pd.to_datetime(sales_df['Date'], format='%d/%m/%Y')
# turn the text dates into real dates (day/month/year)
sales_df['Year'] = sales_df['Date'].dt.year
# a new column with only the year
print(sales_df.groupby('Year')['Weekly_Sales'].sum())
# total sales for each year
```

#### Customer purchase patterns
This dataset has no customer column, so we look at *when* and *what* people buy: holiday weeks vs normal weeks, and the departments that sell the most.

```python
print(sales_df.groupby('IsHoliday')['Weekly_Sales'].mean())
# average weekly sales in normal weeks (False) and holiday weeks (True)
print(sales_df.groupby('Dept')['Weekly_Sales'].sum().sort_values(ascending=False).head(5))
# the 5 departments with the highest total sales
```

#### Store performance
`.sort_values(ascending=False)` sorts from the biggest to the smallest, `.head(5)` shows the first 5 and `.tail(5)` the last 5.

```python
store_sales = sales_df.groupby('Store')['Weekly_Sales'].sum().sort_values(ascending=False)
# total sales of each store, from highest to lowest
print(store_sales.head(5))
# the 5 best stores
print(store_sales.tail(5))
# the 5 weakest stores
print(stores_df['Type'].value_counts())
# how many stores of each type (A, B, C)
print(stores_df.groupby('Type')['Size'].mean())
# the average size of each store type
```

**Retail insights:**
- **Trends:** data from Feb 2010 to Oct 2012. Total sales went up from 2010 (≈ 2.29 billion) to 2011 (≈ 2.45 billion). 2012 looks lower (≈ 2.00 billion) only because it stops in October.
- **Purchase patterns:** holiday weeks sell more on average (≈ 17,036 vs ≈ 15,901 per store-department-week). A few departments (92, 95, 38) bring in the most money.
- **Store performance:** stores 20, 4 and 14 sell the most, and stores 33, 44 and 5 the least. There are 22 type A stores, 17 type B and 6 type C. Type A stores are the biggest (avg ≈ 177,000 sq ft vs ≈ 40,500 for type C), so bigger stores tend to sell more.

```python
reviews_df = pd.read_csv("Women's E-Commerce Clothing Reviews/Womens Clothing E-Commerce Reviews.csv")
# load the reviews CSV (double quotes because the path has an apostrophe)
print(reviews_df.head())
# show the first 5 rows
print(reviews_df.shape)
# number of rows and columns
```

#### Overall satisfaction and sentiment
We measure the sentiment with the star `Rating` (1–5) and `Recommended IND` (1 = would recommend, 0 = would not), because they are the customers' own score of how they feel. Then we read a few reviews to see the words behind the scores. (A tool that scores the text itself, TextBlob, is used in the Ninja exercises.)

`value_counts(normalize=True)` gives each value's share of all rows instead of the count (0.56 = 56%).

```python
print(reviews_df['Rating'].value_counts())
# how many reviews gave each star rating
print(reviews_df['Rating'].value_counts(normalize=True))
# the same as a share of all reviews (0.56 = 56%)
print(reviews_df['Recommended IND'].mean())
# share of reviewers who recommend the item (the mean of 1s and 0s)
```

```python
print(reviews_df['Review Text'].head(5))
# read the text of the first 5 reviews
print(reviews_df[reviews_df['Rating'] == 1]['Review Text'].head(3))
# read 3 reviews with only 1 star, to see what people complain about
```

#### Frequently mentioned topics
We count how many reviews contain some common clothing words. New tools:
- `.dropna()` removes the empty reviews.
- `.str.lower()` turns every review into lower case, so "Fit" and "fit" count the same.
- `.str.contains('fit')` gives `True` for each review that has the word, and `.sum()` counts the `True` values.

```python
texts = reviews_df['Review Text'].dropna().str.lower()
# keep only the reviews that have text, all in lower case
print('fit:', texts.str.contains('fit').sum())
# number of reviews that mention "fit"
print('size:', texts.str.contains('size').sum())
# number of reviews that mention "size"
print('fabric:', texts.str.contains('fabric').sum())
# number of reviews that mention "fabric"
print('color:', texts.str.contains('color').sum())
# number of reviews that mention "color"
print(reviews_df['Department Name'].value_counts())
# which kinds of clothes get the most reviews
```

**Reviews insights:**
- **Sentiment / satisfaction:** most customers are happy: about 56% give 5 stars and 22% give 4 stars, only about 10% give 1–2 stars, and about 82% recommend the item.
- **Topics:** fit and size come up the most (about 40% of reviews mention "fit" and 31% "size"), then color (≈ 27%) and fabric (≈ 18%). Tops and Dresses get the most reviews.
- **Complaints** (from reading the 1-star reviews above): thin or itchy fabric, clothes that don't look like the photo, a bad shape or fit, and sending the item back.

### Part C: Comparison and challenges

| | Retail dataset (structured) | Reviews dataset (unstructured text) |
|---|---|---|
| What it tells us | **what** happens: how much was sold, when and where | **why** it happens: what customers like or dislike |
| Example insight | holiday weeks sell more; type A (big) stores sell the most | customers are mostly satisfied; fit and size are the main problems |
| How easy to analyse | easy: numbers and dates, we can sum, average and group directly | harder: free text must be read, cleaned or turned into numbers first |

**Together:** retail numbers show which stores or departments do well or badly; reviews can explain the reason (for example, many returns because of wrong sizes).

**Challenges of processing unstructured data:**
- **No fixed format:** every review is written differently (length, spelling, slang, emojis), so we can't just sum or average it.
- **Meaning is hard for a computer:** sarcasm, or "not bad", or "I love the color but it doesn't fit" (both positive and negative in one review).
- **Missing and messy text:** 845 reviews have no text and 3,810 have no title; text needs cleaning (lower case, punctuation).
- **Needs extra tools:** to measure sentiment or topics properly we need NLP (text-analysis) methods, which take more time and computing power than simple pandas sums.
- **Simple word counts can mislead:** "fit" is also found inside "fitted" or "outfit", and "small" can be good or bad.

## Exercise 6: Synthetic Product Catalog

### Step 1: Install and import Faker

```python
!pip install faker
# install Faker (if it is already installed, pip just says so)
```

```python
from faker import Faker
# import the Faker class
import random
# Python's random module, for the prices
fake = Faker()
# create a Faker object that makes fake data
```

### Step 2: Create 500 products
Each product is a dictionary. `fake.word()` gives a random word, `.capitalize()` makes its first letter big, `fake.sentence()` gives a random sentence, and `random.uniform(1, 500)` gives a random decimal number between 1 and 500 (`round(..., 2)` keeps 2 decimals).

```python
products = []
# an empty list to collect the products
for i in range(1, 501):
# repeat 500 times, i goes from 1 to 500
    product = {'product_id': i, 'name': fake.word().capitalize(), 'description': fake.sentence(), 'price': round(random.uniform(1, 500), 2)}
    # one product: a unique ID (i), a fake name, a fake sentence and a price between 1 and 500 with 2 decimals
    products.append(product)
    # add the product to the list
```

### Step 3: Make a DataFrame
`.is_unique` is `True` when no value in the column repeats.

```python
catalog_df = pd.DataFrame(products)
# turn the list of products into a DataFrame
print(catalog_df.head())
# show the first 5 products
print(catalog_df.shape)
# should be (500, 4)
print(catalog_df['product_id'].is_unique)
# True means every product ID is different
```
