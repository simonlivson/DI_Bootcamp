# My case: Tracking bird flu (H5N1) through wastewater data in the US, 2024

## Step 1: The story

In spring 2024, a highly pathogenic bird flu virus, H5N1, was spreading among dairy cows in the United States. It was hard to know how big the outbreak really was. On 30 April 2024, CNN reported that US officials admitted it had been difficult to get milk producers to let them test their animals [2].

At the same time, a team from **WastewaterSCAN**, a national wastewater monitoring network run by Stanford University with Emory University, with Verily Life Sciences also involved [2][3], noticed something strange in their data. Influenza A levels in sewage were going **up** at many treatment plants in spring, even though human flu cases were flat or falling [2].

The team investigated, and their study was published in May 2024 in the scientific journal *Environmental Science & Technology Letters* [1]. Later, on 19 September 2024, the US Centers for Disease Control and Prevention (CDC) published its own report on wastewater monitoring for H5 across the country [4].

## Step 2: The role of data analysis

**What data was analyzed**
- Influenza A RNA concentrations measured in wastewater solids from **190 treatment plants** across the US [1].
- Stored (biobanked) samples collected between February and April 2024 from plants in Amarillo and Dallas County (Texas), Forsyth County (North Carolina) and Honolulu (Hawaii), **164 samples** in total [1].
- Records of which industries (like dairy processors) discharge into each sewer system [1].

**Methods used**
1. **Trend and anomaly detection.** They checked each plant's time series for a new rise in influenza A after the flu season ended, and found one at **59 of the 190 plants** [1].
2. **A new targeted measurement.** They built and validated a lab test (RT-PCR) for the H5 gene to separate it from other flu viruses [1].
3. **Comparison with a control group.** They tested four plants with spring increases and one control plant (Honolulu) with no increase and no animal processing [1].
4. **Comparing two measurements.** H5 levels were close to total influenza A levels, suggesting most of the virus was H5 [1].

**Outcomes and decisions**
- H5 was found at all four plants with spring increases and **not** found at the control plant [1].
- In Amarillo, H5 was already present on 25 February, about a week before an unknown illness was reported in Texas cows [3].
- All four positive plants allowed industrial discharges with animal waste, including milk byproducts, so the researchers pointed to animal sources rather than people [1].
- The decision that followed: in May 2024, routine H5 testing started at 193 sites in 41 states, with results on a public dashboard, and CDC added H5 data to its national data pipeline [4].

## Step 3: Impact

**What would have been different without data analysis?**
Without this analysis, the spring rise in influenza A could have been ignored as normal flu noise. Officials would have relied only on testing animals and people, which was limited because many farms did not want to be tested [2]. The February signal would have stayed hidden.

**How data analysis helped**
- It gave an **early warning**. As Professor Alexandria Boehm of Stanford said, "We detected the H5 marker right before there was confirmed highly pathogenic avian influenza in those counties" [2].
- It changed national policy: H5 wastewater testing became routine, and in the CDC's report, 15 of the 24 sites with H5 detections found animal sources nearby, including 8 milk-processing inputs [4].

**Limits I noticed**
The analysis could not show which animal the virus came from, and the test detects H5 in general, not only H5N1 [3][4]. So the data pointed investigators in the right direction, but it still had to be checked with field work.

## Step 4: Conclusion

This story shows that data analysis can protect public health. The WastewaterSCAN team started with a simple time series of flu levels, but by asking why the trend did not match the normal pattern, they found evidence of bird flu in animal waste. They used clear methods: spotting anomalies, building a new measurement, comparing with a control and linking results to data about local industries.

The result was a real decision at the national level: routine H5 wastewater testing in 41 states and a public dashboard. For me, the main lesson is that good data analysis can see a problem earlier and more cheaply than traditional methods, and that being honest about the limits of the data is part of doing the analysis well.
